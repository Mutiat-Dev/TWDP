### A data-driven initiative that identifies gaps in waste collection and disposal across states, using public complaints to uncover trends, evaluate service delivery, and support evidence-based solutions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


file_id = '1QIsWdo7MBxqTUDrBxD7jiorDdaIHJoS6'
url = f'https://drive.google.com/uc?export=download&id={file_id}'

waste_df = pd.read_csv(url)

waste_df.info()
waste_df.head(10)
```

### Reports by State
This is to see which states have the highest number of reported waste issues

```python
state_counts = waste_df['state'].value_counts()

plt.figure(figsize=(8, 6))
state_counts.plot(kind='bar', color='green')
plt.title('Number of Waste Reports by State')
plt.xlabel('State')
plt.ylabel('Number of Reports')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Distinct waste type
This is to show the unique waste type reported in the dataset

```python
wst_type = waste_df['waste_type'].unique()
print(wst_type)

waste_df[['waste_type', 'urgency_level']]


```

### Degree of urgency
Thisis to show the level of urgency for each waste type reported in the dataset to show priority in case of government intervention

```python
plt.figure(figsize=(6, 6))
plt.barh(waste_df['waste_type'], waste_df['urgency_level'], color='skyblue')
plt.xlabel('Waste Type')
plt.ylabel('Urgency Level')
plt.title('Waste Type vs. Urgency Level')

plt.show()
```

### Preferred communication channel for updates
This is to show the behaviour of respondent to communication channels reported in the dataset

```python
p_c_c = waste_df['update_preference'].value_counts()
print(p_c_c)

# p_c_c.unique()
plt.plot(waste_df['update_preference'])
plt.title('Preferred communication channel for updates')
plt.show()

```

### Future Interest
This to show the percentage of respondent that'd like to pay for or use an app for waste management in the future

```python
waste_df['future_app_interest'] = waste_df['future_app_interest'].replace({
    'Yes. I want a faster, easier way to report.': 'Yes',
    'Maybe. I would need to see how effective it is first.': 'Maybe',
    'No. I prefer using WhatsApp or phone calls.': 'No'
})
app = waste_df['future_app_interest'].value_counts()

plt.pie(app, labels=app.index, autopct='%1.1f%%', startangle=90)
plt.title('App Preference')
plt.tight_layout()
plt.show()
```

### Analysis of Community Impact
Community_impact contain multiple values (e.g., "Bad smell, Health risk"), we want to split and count individual impacts

```python
all_impacts = waste_df['community_impact'].dropna().str.split(', ')
exploded_impacts = all_impacts.explode()
impact_counts = exploded_impacts.value_counts()

plt.figure(figsize=(10, 5))
impact_counts.plot(kind='barh', color='tomato')
plt.title('Most Common Community Impacts of Waste')
plt.xlabel('Count')
plt.ylabel('Impact Type')
plt.tight_layout()
plt.show()
```
